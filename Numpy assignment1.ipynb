```python
import numpy as np

#Create a 1D NumPy array for Week 1

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:")
print(temperatures_w1)
```

```python
# Array properties

print("Shape:", temperatures_w1.shape)

print("Data Type:", temperatures_w1.dtype)

print("Number of Elements:", temperatures_w1.size)
```

```python
# Array Operations

#  Celsius to Fahrenheit
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:")
print(temperatures_f)

# Maximum temperature
print("Maximum Temperature:", temperatures_w1.max())

# Minimum temperature
print("Minimum Temperature:", temperatures_w1.min())

# Mean temperature
print("Mean Temperature:", temperatures_w1.mean())
```

```python
#  Array Slicing and Indexing

# First three days
print("First Three Days:", temperatures_w1[:3])

# Weekend temperatures
print("Weekend Temperatures:", temperatures_w1[-2:])

# Middle three days
print("Middle Three Days:", temperatures_w1[2:5])
```

```python
# 2D NumPy Array

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures for Both Weeks:")
print(temperatures)
```

```python
#  2D array properties

print("Shape:", temperatures.shape)

print("Data Type:", temperatures.dtype)

print("Total Number of Elements:", temperatures.size)
```

```python
# Extract temperatures of each week and weekends

# Week 1 temperatures
print("Week 1:", temperatures[0])

# Week 2 temperatures
print("Week 2:", temperatures[1])

# Weekend temperatures for both weeks
print("Weekend Temperatures:")
print(temperatures[:, -2:])
```

```python
# Pandas Series

import pandas as pd

# Create a Pandas Series with custom indices
marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print("Student Marks:")
print(marks)
```

```python
# Indexing and Slicing

# 1st rank student's mark using integer position
print("Mark of 1st rank student:", marks.iloc[0])

# Marks of the top 3 ranks using labels
print("\nMarks of top 3 ranks:")
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])

# 3rd rank student's mark using iloc
print("\nMark of 3rd rank student:", marks.iloc[2])

# Marks greater than 90
print("\nRanks with marks greater than 90:")
print(marks[marks > 90])
```

```python
# Manipulating Series

# Modifying the mark of the 1st rank student to 100
marks.loc['Rank1'] = 100

print("Updated Marks:")
print(marks)
```

```python
# Remove the last-ranked student's entry

marks = marks.drop('Rank5')

print("Marks after removing the last rank:")
print(marks)
```

```python
# CGPA by dividing each mark by 10

cgpa = marks / 10

print("CGPA:")
print(cgpa)
```

```python
# Pandas DataFrame
transactions = pd.DataFrame({
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],

    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],

    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],

    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
})

print(transactions)
```

```python
# The complete DataFrame
print("Transactions DataFrame:")
print(transactions)

# The first five rows
print("\nFirst Five Rows:")
print(transactions.head())

# The last five rows
print("\nLast Five Rows:")
print(transactions.tail())

# The shape
print("\nShape:")
print(transactions.shape)

# Column names
print("\nColumn Names:")
print(transactions.columns)

# Data types
print("\nData Types:")
print(transactions.dtypes)
```

```python
# ProductCategory and Amount columns
print(transactions[['ProductCategory', 'Amount']])
```

```python
# The last 3 columns
print("Last 3 Columns:")
print(transactions.iloc[:, -3:])
```

```python
# Filter transactions
filtered_transactions = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(filtered_transactions)
```

```python
print(transactions['ProductCategory'].value_counts())
```

```python
print(transactions['Region'].unique())
```

```python
# Group by Region and the mean Amount of each region
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# Modifying the Amount for TransactionID 102
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

print(transactions)
```

```python
# Discount column
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

```python
# Removing the row with TransactionID 109
transactions = transactions.drop(
    transactions[transactions['TransactionID'] == 109].index
)

print(transactions)
```

```python
# Deleting the Discount column
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```
