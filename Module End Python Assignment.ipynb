```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
import warnings
warnings.filterwarnings('ignore')

file_path = 'social_media_engagement_5000.csv'
df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
display(df.head())
```

```python
print("First 5 rows:")
display(df.head())

print("Last 5 rows:")
display(df.tail())

print("Dataset shape:", df.shape)
print("Column names:", df.columns.tolist())

print("\\nDataFrame information:")
df.info()

print("\\nData types:")
display(df.dtypes)

# Date column to datetime 
df['posted_at'] = pd.to_datetime(df['posted_at'], format='%d-%m-%Y', errors='coerce')

# Converting boolean-like verified column to boolean
df['is_verified'] = df['is_verified'].astype('bool')

print("\\nDate range:", df['posted_at'].min(), "to", df['posted_at'].max())

# NumPy array operations using the likes column
likes_array = df['likes'].dropna().to_numpy()
print("NumPy likes array (first 10):", likes_array[:10])
print("Total likes:", np.sum(likes_array))
print("Average likes:", np.mean(likes_array))
print("Maximum likes:", np.max(likes_array))
```

```python
# Missing values
print("Missing values by column:")
display(df.isnull().sum().to_frame('Missing_Count'))

print("Total missing values:", df.isna().sum().sum())

# Removing exact duplicate rows
exact_duplicates = df.duplicated().sum()
print("Exact duplicate rows:", exact_duplicates)
df = df.drop_duplicates().copy()

# Filling numeric missing values with median
numeric_fill_columns = ['age', 'likes', 'comments', 'shares']
for col in numeric_fill_columns:
    df[col] = df[col].fillna(df[col].median())

# Filling categorical missing values with mode
for col in ['gender', 'sentiment']:
    mode_value = df[col].mode(dropna=True)[0]
    df[col] = df[col].fillna(mode_value)

# Removing extra spaces and lowercase
categorical_columns = ['gender', 'country', 'post_type', 'post_category',
                       'device_type', 'sentiment', 'hashtags']


for col in categorical_columns:
    df[col] = df[col].astype(str).str.strip().str.lower()

numeric_check_columns = ['age', 'likes', 'comments', 'shares',
                         'watch_time_sec', 'impression_count',
                         'follower_count', 'engagement_rate']
display(df[numeric_check_columns].describe().T)



for col in ['age', 'likes', 'comments', 'shares', 'watch_time_sec',
            'impression_count', 'follower_count', 'engagement_rate']:
    print(f"Negative values in {col}:", (df[col] < 0).sum())

print("\\nMissing values after cleaning:")
display(df.isna().sum().to_frame('Missing_After_Cleaning'))
```

```python
# Extracting hashtag count from each post
df['hashtag_count'] = df['hashtags'].apply(
    lambda x: sum(1 for tag in str(x).split() if tag.startswith('#'))
)

# Creating engagement score as total likes + comments + shares
df['engagement_score'] = df['likes'] + df['comments'] + df['shares']

# Extracting calendar features from posted_at
df['post_date'] = df['posted_at'].dt.date
df['post_day'] = df['posted_at'].dt.day_name()
df['post_month'] = df['posted_at'].dt.month_name()
df['post_year'] = df['posted_at'].dt.year


df['log_impressions'] = np.log1p(df['impression_count'])
df['log_followers'] = np.log1p(df['follower_count'])

display(df[['hashtags', 'hashtag_count', 'engagement_score',
            'post_day', 'post_month', 'post_year']].head())
```

```python
print("Summary statistics:")
display(df.describe(include='all').T)

for col in ['gender', 'country', 'post_type', 'post_category',
            'device_type', 'sentiment']:
    print(f"\\nDistribution for {col}:")
    display(df[col].value_counts().to_frame('count'))

print("\\nUnique values in post_type:", df['post_type'].unique())
print("Number of unique countries:", df['country'].nunique())
print("Number of unique post categories:", df['post_category'].nunique())

# Correlation matrix for numeric columns
numeric_df = df.select_dtypes(include=np.number)
correlation_matrix = numeric_df.corr()
display(correlation_matrix.round(3))

# Groupby summaries
print("Average likes by post type:")
display(df.groupby('post_type')['likes'].mean().sort_values(ascending=False).to_frame('average_likes'))

print("Average impressions by country:")
display(df.groupby('country')['impression_count'].mean().sort_values(ascending=False).to_frame('average_impressions'))
```

```python
# Grouped summary by post type
post_type_summary = df.groupby('post_type').agg(
    post_count=('post_id', 'count'),
    average_likes=('likes', 'mean'),
    average_comments=('comments', 'mean'),
    average_shares=('shares', 'mean'),
    average_engagement_rate=('engagement_rate', 'mean')
).sort_values('average_engagement_rate', ascending=False)
display(post_type_summary.round(3))

# Grouped summaries by category, country and sentiment
category_summary = df.groupby('post_category').agg(
    posts=('post_id', 'count'),
    avg_engagement=('engagement_rate', 'mean'),
    avg_likes=('likes', 'mean')
).sort_values('avg_engagement', ascending=False)

country_summary = df.groupby('country').agg(
    posts=('post_id', 'count'),
    avg_engagement=('engagement_rate', 'mean')
).sort_values('avg_engagement', ascending=False)

sentiment_summary = df.groupby('sentiment').agg(
    posts=('post_id', 'count'),
    avg_engagement=('engagement_rate', 'mean'),
    avg_watch_time=('watch_time_sec', 'mean')
).sort_values('avg_engagement', ascending=False)

display(category_summary.round(3))
display(country_summary.round(3))
display(sentiment_summary.round(3))

# MERGE: attach country-level average engagement to each post
df_with_country_avg = df.merge(
    country_summary[['avg_engagement']],
    left_on='country', right_index=True, how='left'
)
display(df_with_country_avg[['country', 'engagement_rate', 'avg_engagement']].head())

# CONCAT: split data into two parts and combine them back (demonstrates row-wise concat)
first_half = df.iloc[:len(df)//2]
second_half = df.iloc[len(df)//2:]
df_concat = pd.concat([first_half, second_half], ignore_index=True)
print("Rows after concatenating the two parts:", len(df_concat))

# JOIN: join category summary back to the main data by category
df_joined = df.set_index('post_category').join(
    category_summary[['avg_engagement']].rename(
        columns={'avg_engagement': 'category_avg_engagement'}
    )
)
display(df_joined[['post_type', 'engagement_rate', 'category_avg_engagement']].head())
```

```python
metrics = ['likes', 'comments', 'shares', 'watch_time_sec',
           'engagement_rate', 'follower_count']

statistics = pd.DataFrame(index=metrics)
statistics['mean'] = df[metrics].mean()
statistics['median'] = df[metrics].median()
statistics['mode'] = df[metrics].mode().iloc[0]
statistics['standard_deviation'] = df[metrics].std()
statistics['variance'] = df[metrics].var()
statistics['25th_percentile'] = df[metrics].quantile(0.25)
statistics['50th_percentile'] = df[metrics].quantile(0.50)
statistics['75th_percentile'] = df[metrics].quantile(0.75)
statistics['90th_percentile'] = df[metrics].quantile(0.90)
statistics['skewness'] = df[metrics].skew()
statistics['kurtosis'] = df[metrics].kurtosis()

display(statistics.round(3))
```

```python
# Calculate insights directly from the cleaned data
type_insight = df.groupby('post_type')['engagement_rate'].mean().sort_values(ascending=False)
category_insight = df.groupby('post_category')['engagement_rate'].mean().sort_values(ascending=False)
country_insight = df.groupby('country')['engagement_rate'].mean().sort_values(ascending=False)
age_insight = df.groupby(pd.cut(df['age'], bins=[0,19,29,39,49,59,100],
                                labels=['13-19','20-29','30-39','40-49','50-59','60+']),
                         observed=False)['engagement_rate'].mean()
verified_insight = df.groupby('is_verified')['engagement_rate'].mean()
device_insight = df.groupby('device_type')['watch_time_sec'].mean().sort_values(ascending=False)
sentiment_insight = df.groupby('sentiment')['engagement_rate'].mean().sort_values(ascending=False)
weekday_insight = df.groupby('post_day')['engagement_rate'].mean().sort_values(ascending=False)

print("1. Highest-engagement post type:", type_insight.index[0],
      "| Average engagement:", round(type_insight.iloc[0], 3))
print("2. Best-performing category:", category_insight.index[0],
      "| Average engagement:", round(category_insight.iloc[0], 3))
print("3. Country with highest average engagement:", country_insight.index[0],
      "| Average engagement:", round(country_insight.iloc[0], 3))
print("\\n4. Average engagement by age group:")
display(age_insight.round(3).to_frame('average_engagement'))
print("5. Verified vs unverified average engagement:")
display(verified_insight.round(3).to_frame('average_engagement'))
print("6. Device with highest average watch time:", device_insight.index[0],
      "| Average watch time (seconds):", round(device_insight.iloc[0], 2))
print("7. Sentiment with highest average engagement:", sentiment_insight.index[0],
      "| Average engagement:", round(sentiment_insight.iloc[0], 3))
print("8. Day with highest average engagement:", weekday_insight.index[0],
      "| Average engagement:", round(weekday_insight.iloc[0], 3))

print("\\nPosting time of day: Cannot be analyzed because the posted_at field contains dates only, not clock times.")
```
